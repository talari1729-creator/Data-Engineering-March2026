```python
titanic=df
```

```python
# survival count
plt.figure(figsize=(6,4))
sns.countplot(data=titanic,x='sex')
plt.title('Survival count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=titanic,x='class')
plt.title(" survival class count")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.histplot(data=titanic,x='age',kde=True)
plt.title("fare distribution")

plt.show()
## kde=True

# kde = Kernel Density Estimation

# It adds a smooth curve over the histogram.
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(data=titanic,y='class',x='age')
plt.title("age by passenger class")
plt.show()
```

```python
plt.figure(figsize=(5,3))
sns.boxplot(data=titanic,x='class',y='age')
plt.title("age by passenger class")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(data=titanic,x='survived',y='age')
plt.title('age vs survival')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=titanic,x='age',y='fare',hue='survived')
plt.title('age vs fare')
plt.show()
# hue="survived"
# → split each bar by survival status using different colors
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(data=titanic, x="age", y="fare", hue="survived")
plt.title("Age vs Fare")
plt.show()
```

```python
plt.figure(figsize=(4,4))
sns.heatmap(data,annot=True,linewidths=2,linecolor='g')
#annot means annotation.
#It writes the actual numbers inside each box.
```

```python
plt.figure(figsize=(6,4))
sns.swarmplot(data=titanic, x="class", y="age")
plt.title("Class vs Age")
plt.show()
```
