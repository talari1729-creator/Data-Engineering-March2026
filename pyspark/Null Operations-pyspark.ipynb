```python
from pyspark.sql import SparkSession 
from pyspark.sql.functions import * 
spark=SparkSession.builder.appName("spark_null_functions").getOrCreate()

data=[
    (18,"raajan",50000,None),
    (None,"jaswanth",25000,"police"),
    (15,"niranjan",None,"ISS"),
    (13,"kousik",40000,None),
    (22,None,50000,"SBI PO"),
    (21,"teja",None,"teacher")
]
df=spark.createDataFrame(data,["jersey","name","salary","department"])

display(df.show())
```

```python
# 1. isNull()
# Returns True if value is NULL
df.select("jersey",col("name").isNull().alias("name is null")).show()

df.filter(col("name").isNull().alias("name is null")).show()
```

```python
# 2. isNotNull()
# Returns True if value is NOT NULL
df.select("jersey",col("name").isNotNull().alias("name is not null")).show()
```

```python
df.select("name",col("department").isNull().alias("department is null")).show()
```

```python
# 3. filter() with isNull()
# Get rows containing NULL values
df.filter(col("name").isNull()).show()
```

```python
# 4. filter() with isNotNull()
# Get rows containing non-NULL values
df.filter(col("name").isNotNull()).show()
```

```python
# 5. fillna() - Replace NULL with constant value
df.fillna("Bhanu",subset=["name"]).show()
```

```python
df.show()
```

```python
# 6. fillna() Multiple Columns
df.fillna({
    "name":"Bhanu",
    "jersey":35,
    "salary":70000,
    "department":"unpredictable"
}).show()
```

```python
df.show()
```

```python
# 7. na.fill()
# Same as fillna()
df.na.fill("missing",subset=["name"]).show()
```

```python
# 8. dropna()
# Remove rows having NULL values
df.dropna().show()
```

```python
# 9. na.drop()
# Same as dropna()
df.na.drop().show()
```

```python
# 10. dropna(how="any")
# Drop row if ANY column is NULL
df.dropna(how="any").show()
```

```python
# 11. dropna(how="all")
# Drop row if ALL columns are NULL
df.dropna(how="all").show()
```

```python
# 12. dropna(subset=[])
# Drop rows if specified column contains NULL
df.dropna(subset=["name"]).show()
```

```python
df.dropna(subset=["name","department"]).show()
```

```python
# 13. dropna(thresh=3)
# Keep rows having at least 3 non-null values
df.dropna(thresh=3).show()
```

```python
# 14. coalesce()
# Returns first non-null value
df.withColumn(
    "dept_final",
    coalesce(col("department"),lit("not assigned"))
).show()
```

```python
# 15. ifnull() / nvl()
# Replace NULL with another value
df.selectExpr(
    "jersey",
    "nvl(name,'bhanu')as employee_name" # bhanu must within single quants# 16. nvl2()
# If value not null return first expression else second
).show()
```

```python
# 16. nvl2()
# If value not null return first expression else second
df.selectExpr(
    "jersey",
    "nvl2(name,'present','missing')as status"
).show()
```

```python
# 17. nullif()
# Returns NULL if two expressions are equal
# returns null values as Null and also what should be want to get null
df.selectExpr(
    "name",
    "nullif(department,'police')as dept"
).show()
```

```python
# 18. when() + otherwise()
# Conditional NULL handling
# when col salary is null,filled with some value other wise get same value salary
df.withColumn(
    "salary_new",
    when(col("salary").isNull(),40000).otherwise(col("salary"))
).show()
```

```python
# 19. count()
# Counts only NON-NULL values
df.select(count("salary").alias("no.of.non.null")).show()
```

```python
# 20. Count NULL Values Per Column
# c reads all columns names
df.select([
    count(when(col(c).isNull(), c)).alias(c + "_null_count")
    for c in df.columns
]).show()
```

```python
df.select([
    count(when(col(c).isNull(),c)).alias(c+"_null_count") for c in df.columns
]).show()
```

```python
# 21. Replace NULL using SQL Expression
df.selectExpr(
    "jersey",
    "coalesce(name,'unknown')as name",
    "coalesce(salary,0)as salary"
).show()
```

```python
# 22. Find Rows Having Any NULL Column by concate
df.filter(
    col("name").isNull() |
    col("salary").isNull()
).show()
```

```python
# 23. Find Rows Having No NULL Columns by concate
df.filter(
    col("name").isNotNull()|
    col("salary").isNotNull()
).show()
```

```python
# 24. Create Null Indicator Columns
df.withColumn(
    "salary_missing",
    when(col("salary").isNull(),1).otherwise(0)
).show()
```

```python
display(
    df.select([
        (count(when(col(c).isNull(), c)) / count(lit(1))).alias(c + "_null_pct")
        for c in df.columns
    ])
)
```
