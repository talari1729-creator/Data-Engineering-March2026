```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark=SparkSession.builder.appName("spark dtaframe").getOrCreate()

data=[(101,'raajan',25,50000,'IT'),
      (102,'janardhan',21,60000,'HR'),
      (103,'narasimhulu',45,70000,'ceo'),
      (104,'lakshmi',40,80000,'manager'),
      (105,'venkatesh',39,58000,'IT'),
      (106,'aachari',24,55000,'HR'),
      (107,'muni',35,47000,'manager'),
      (108,'ram',30,59000,'ceo'),
      (109,'prasad',40,75000,'manager'),
      (110,'jaswanth',33,38000,'HR')]
columns=["emp_id","name","age","salary","dept"]
df=spark.createDataFrame(data,columns)
df.show()
```

```python
df.select("name","salary").show()
```

```python
df.select("name","dept").show(5)
```

## Aithmetic operations


```python
# addition
df.select("name","salary",(col("salary")+5000).alias("bonus")).show()
```

```python
# subtraction
df.select("name","salary",(col("salary")-5000).alias("decrement")).show()
```

```python
# multiplication
df.select("name","salary",(col("salary")*2).alias("100% hike")).show()
```

```python
# division
df.select("name","salary",(col("salary")/2).alias("50%decrement")).show()
```

```python
# modulus
df.select("name","salary",(col("salary")%7).alias("remainder to the salary")).show()
```

```python
# exponentation
df.select("name","salary",(col("salary")**2).alias("power to that salary")).show()
```

# Comparision operations

```python
df.select("name","salary",(col("salary")<50000)).show()
```

# if you use select it shows booleans
#  if you use filter it shows details

```python
df.filter(col("salary")<=50000).show()
df.filter(col("salary")<50000).show()
```

```python
df.filter(col("salary")>=50000).show()
df.filter(col("salary")>50000).show()
```

```python
df.filter(col("salary")==50000).show()
df.filter(col("salary")!=50000).show()
```

# Logical operations

```python
# and operator
df.filter(
    (col("salary")<50000) &
    (col("dept")=="HR")).show()

df.filter(
    (col("salary")>50000) &
    (col("dept")=="HR")).show()
```

```python
# or operator
df.filter(
    (col("salary")<60000)|
    (col("dept")=="md")
).show()
```

```python
# not operator
df.filter(
    ~(col("salary")>=50000)
).show()
```
