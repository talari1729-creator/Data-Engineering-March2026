```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark = SparkSession.builder.appName("RetailTransformations").getOrCreate()

# Retail dataset 
data = [
    (1, "2026-01-05", "Bangalore", "Laptop", "Electronics", 2, 55000),
    (2, "2026-01-10", "Hyderabad", "Mobile", "Electronics", 3, 25000),
    (3, "2026-02-15", "Chennai", "Shoes", "Fashion", 5, 3000),
    (4, "2026-02-20", "Bangalore", "TV", "Electronics", 1, 45000),
    (5, "2026-03-01", "Pune", "Shirt", "Fashion", 4, 1200),
    (6, "2026-03-12", "Hyderabad", "Rice Bag", "Grocery", 10, 900),
    (7, "2026-04-05", "Chennai", "Headphones", "Electronics", 6, 2000),
    (8, "2026-04-18", "Pune", "Jeans", "Fashion", 3, 2500),
    (9, "2026-05-08", "Bangalore", "Oil Packet", "Grocery", 8, 180),
    (10, "2026-05-20", "Hyderabad", "Watch", "Fashion", 2, 5000)
]

columns = ["order_id", "order_date", "city", "product", "category", "quantity", "price"]
df=spark.createDataFrame(data,columns)

# Convert string date to date type
df=df.withColumn("order_date",to_date(col("order_date")))
df.show()
df.printSchema()
```

```python
df.printSchema()
```

```python
df_select=df.select("order_id","order_date","city","product","category")
display(df_select.show())
```

```python
display(df_select)
```

```python
# 2. filter transformation
df_filter=df.filter(col("category")=="Electronics")
df_filter.show()
```

```python
# 3. withColumn transformation
df_amount=df.withColumn("total_amount",col("quantity")*col("price"))
df_amount.show()
```

```python
# 4. Extract year, month, day from date
df_date=df.withColumn("year",year(col("order_date")))\
          .withColumn("month",month(col("order_date")))\
          .withColumn("day",dayofmonth(col("order_date")))

df_date.show()              
```

```python
# 6. filter date range
df_date_filter=df.filter(
    (col("order_date")>="2026-02-01")&
    (col("order_date")<="2026-04-30")
)
df_date_filter.show()
```

```python
## month,category total sales,order by month
df_category_sales=df.withColumn("total_amount",col("quantity")*col("price"))\
    .withColumn("month",month(col("order_date")))\
    .groupBy("month")\
    .agg(sum("total_amount").alias("total_sales"))
df_category_sales.show()
# 8. group by category, total sales
#df_category_sales=df.withColumn("total_amount",col("quantity")*col("price"))\
#    .groupBy("category")\
#    .agg(sum("total_amount").alias("total_sales"))
#df_category_sales.show()                  
```

```python
# 12. drop transformation
df_drop=df.drop("price")
df_drop.show()
```

```python
df_rename=df.withColumnRenamed("city","store_city")
df_rename.show()
```

```python
# 14. distinct transformation
df_distinct_city=df.select("city").distinct()
df_distinct_city.show()
```

```python
# 15. sort transformation
df_sort=df.sort(col("order_date").desc())
df_sort.show()
```

```python
# 16. limit transformation
df_limit=df.limit(5)
df_limit.show()
```

```python
# 17. dropDuplicates transformation
df_drop_duplicates=df.dropDuplicates(["city"])
df_drop_duplicates.show()
```

```python
df_orderby_month=df.withColumn("total_sales",col("quantity")*col("price"))\
    .withColumn("month",month("order_date"))\
    .groupBy("month","category")\
    .agg(sum("total_sales").alias("total_sales"))\
    .orderBy("month")
df_orderby_month.show()
```

```python
df_month_category_sales = df.withColumn("total_sales", col("quantity") * col("price")) \
    .withColumn("month", month(col("order_date"))) \
    .groupBy("month", "category") \
    .agg(sum("total_sales").alias("total_sales")) \
    .orderBy("month")

display(df_month_category_sales)
```
