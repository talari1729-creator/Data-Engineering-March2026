```python
# Masking data 
data=[
    ("177236474698698",),
    ("48983705705737",),
    ("4597049570938409570",)
]
df=spark.createDataFrame(data,["Id number"])
df.show()

def mask_num(number):
    if number is None :
        return "unknown"
    return "*******"+number[-4:]

mask_udf=udf(mask_num,StringType())

result=df.withColumn(
        "masked number",
        mask_udf("Id number")
    )

result.show()
```

```python
# masking first and last numbers of aadhar
data=[
    ("182736405847",),
    ("948573748374",),
    ("484856573992",)
]
df=spark.createDataFrame(data,["aadhar"])
df.show()

def mask_aadhar(aadhar):
    if aadhar is None: 
        return "unknown"
    n=len(aadhar)
    start=(n-4)//2
    middle=aadhar[start:start+4]
    return "X"*start+middle+"X"*(n-start-4)
    
    
aadhar_mask=udf(mask_aadhar,StringType())

result=df.withColumn(
    "masked aadhar",
    aadhar_mask("aadhar")
)

result.show()
```

```python
# masking middle numbers of aadhar 
data=[
    ("837465928460",),
    ("967294857264",),
    ("907485625483",)
]

df=spark.createDataFrame(data,["adhar"])
df.show()

def masked_adhar(adhar):
    if adhar is None: 
        return "unknown"
    # n=len(adhar)
    # start=(n-4)//2
    # middle=adhar[start:start+4]
    return adhar[:4]+"x"*(4)+adhar[8:]
mask_adhar=udf(masked_adhar,StringType())

result=df.withColumn(
    "masking_column",
    mask_adhar("adhar")
)

result.show()
```

```python
# encrypt using fernet

from cryptography.fernet import Fernet
from pyspark.sql.types import *
data = [
    (101, "Alice",   "alice@gmail.com",   "9876543210"),
    (102, "Bob",     "bob@gmail.com",     "9123456780"),
    (103, "Charlie", "charlie@gmail.com", "9988776655"),
    (104, "David",   "david@gmail.com",   "9871234567"),
    (105, "Eva",     "eva@gmail.com",     "9012345678")
]

columns = ["emp_id", "name", "email", "phone"]

df = spark.createDataFrame(data, columns)
df.show()

key=Fernet.generate_key()
fernet=Fernet(key)

# encrypt the phone

def encrypt_phone(phone):
    if phone is None :
        return "unknown"
    return fernet.encrypt(phone.encode()).decode()

encrypt_phone_udf=udf(encrypt_phone,StringType())
result=df.withColumn(
    "encrypted_phone",
    encrypt_phone_udf("phone")
)
result.show()
```

```python
# Feature transformations 
from pyspark.sql import SparkSession 
from pyspark.sql.functions import udf,col
from pyspark.sql.types import DoubleType
import math
```

```python
spark=SparkSession.builder.appName("feature_transformations").getOrCreate()

data = [
    (1, 1000.0, 2.0, 4.0, 25.0, 3.0),
    (2, 5000.0, 4.0, 9.0, 36.0, 5.0),
    (3, 10000.0, 5.0, 16.0, 49.0, 7.0),
    (4, 50000.0, 10.0, 25.0, 64.0, 9.0),
    (5, 100000.0, 20.0, 36.0, 81.0, 11.0)
]

columns=["id","salary","distance","age","sales","score"]

df=spark.createDataFrame(data,columns)
df.show()
```

```python
#single udf for single feature transformations
def reciprocal_salary(salary):
    if salary is None:
        return None
    return 1/salary 

reci_udf=udf(reciprocal_salary,DoubleType())

result=df.withColumn(
    "reciprocal_sal",
    reci_udf("salary")
)
result.show()
```

```python
# single udf for all feature transformations 
def feature_transforma(value,transform_type):
    if value is None:
        return None
    value=float(value)

    if transform_type=="log":
        if value<=0:
            return None
        return math.log(value)
    elif transform_type=="reciprocal":
        if value == 0:
            return None 
        return 1/value 
    elif transform_type=="square":
        return value**2
    elif transform_type=="square_root":
        if value <0:
            return None 
        return math.sqrt(value)
    elif transform_type=="cube":
        return value **3
    elif transform_type=="power":
        return value**4
    elif transform_type=="custom":
        return value/100
    else:
        return value
    

feature_transform_udf=udf(feature_transforma,DoubleType())

result=df.withColumn(
    "salary_log",
    feature_transform_udf(col("salary"),udf(lambda:"log")())
)
result.show()

result1=df.withColumn(
    "salary_square",
    feature_transform_udf(col("salary"),udf(lambda:"square")())
)
result1.show()


    
```

```python
## Apply UDF on specific columns
from pyspark.sql.functions import lit
def feature_transforma(value,transform_type):
    if value is None:
        return None
    value=float(value)

    if transform_type=="log":
        if value<=0:
            return None
        return math.log(value)
    elif transform_type=="reciprocal":
        if value == 0:
            return None 
        return 1/value 
    elif transform_type=="square":
        return value**2
    elif transform_type=="square_root":
        if value <0:
            return None 
        return math.sqrt(value)
    elif transform_type=="cube":
        return value **3
    elif transform_type=="power":
        return value**4
    elif transform_type=="custom":
        return value/100
    else:
        return value
    

feature_transform_udf=udf(feature_transforma,DoubleType())

df_transformed=(
    df
    .withColumn("salary_log",feature_transform_udf(col("salary"),lit("log")))
     .withColumn("distance_reciprocal", feature_transform_udf(col("distance"), lit("reciprocal")))
    .withColumn("age_square", feature_transform_udf(col("age"), lit("square")))
    .withColumn("sales_square_root", feature_transform_udf(col("sales"), lit("square_root")))
    .withColumn("score_power", feature_transform_udf(col("score"), lit("power")))
    .withColumn("score_custom", feature_transform_udf(col("score"), lit("custom")))
)

print("transformed data")
df_transformed.show()
```
