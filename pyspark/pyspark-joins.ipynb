# JOINS

```python
# employee data
from pyspark.sql.functions import *
emp_data=[
    (1,"roopa",101),
    (2,"vennela",102),
    (3,"vinitha",103),
    (4,"neeraja",104),
    (5,"krishna",None)
]
emp_df=spark.createDataFrame(emp_data,["emp_id","emp_name","dept_id"]) 
#department data
dept_data=[
    (101,"MANAGER"),
    (102,"IT"),
    (103,"HR"),
    (105,"CEO")
]
dept_df=spark.createDataFrame(dept_data,["dept_id","dept_name"])

emp_df.show()
dept_df.show()

```

```python
# inner join
df_inner=emp_df.join(dept_df,"dept_id",'inner')
df_inner.show()
```

```python
# left join 
df_left=emp_df.join(dept_df,"dept_id",'left')
df_left.show()
```

```python
# right join 
df_right=emp_df.join(dept_df,"dept_id","right")
df_right.show()
```

```python
# full join 
df_full=emp_df.join(dept_df,"dept_id","full")
df_full.show()
```

```python
# lest semi   (common dept_id's)
df_left_semi=emp_df.join(dept_df,"dept_id","left_semi")
df_left_semi.show()
```

```python
# left_anti (non common dept_id's)
df_left_anti=emp_df.join(dept_df,"dept_id","left_anti")
df_left_anti.show()
```

```python
# cross join 
df_cross=emp_df.crossJoin(dept_df)
df_cross.show()
```
