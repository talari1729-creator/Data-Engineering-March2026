```python
df.show()
```

```python
# Q1. Extract the year from Sales_Date column.
from pyspark.sql.functions import year
df.select("Sales_Date",year("Sales_Date").alias("year")).show()
```

```python
# Q2. Extract the month from Sales_Date column.
from pyspark.sql.functions import month
df.select("Sales_date",month("Sales_Date").alias("month")).show()
```

```python
# Q3. Extract the day of month from Sales_Date column.
from pyspark.sql.functions import dayofmonth
df.select("Sales_Date",dayofmonth("Sales_Date").alias("day of month")).show()
```

```python
# Q4. Find the day of week for each Sales_Date.
from pyspark.sql.functions import dayofweek
df.select("Sales_Date",dayofweek("Sales_Date").alias("day of week")).show()
```

```python
# Q5. Find the day number within the year.
from pyspark.sql.functions import dayofyear
df.select("Sales_Date",dayofyear("Sales_Date").alias("day of year")).show()
```

```python
# Q6. Find the week number of the year.
from pyspark.sql.functions import weekofyear
df.select("Sales_Date",weekofyear("Sales_Date").alias("week of year")).show()
```

```python
# Q7. Determine the quarter for each Sales_Date.
from pyspark.sql.functions import quarter
df.select("Sales_Date",quarter("Sales_Date").alias("quarter")).show()
```

```python
# Q8. Find the last day of the month for Sales_Date.
from pyspark.sql.functions import last_day
df.select("Sales_Date",last_day("Sales_Date").alias("last day of month")).show()
```

```python
# Q9. Add 3 months to Sales_Date.
from pyspark.sql.functions import add_months
df.select("Sales_Date",add_months("Sales_Date",3).alias("add 3 months")).show()
```

```python
# Q10. Subtract 2 months from Sales_Date.
from pyspark.sql.functions import add_months
df.select("Sales_Date",add_months("Sales_Date",-2).alias("subtract 2 months")).show()
```

```python
# Q11. Add 15 days to Sales_Date.
from pyspark.sql.functions import date_add
df.select("Sales_Date",date_add("Sales_Date",15).alias(" add 15 days")).show()
```

```python
# Q12. subtract 20 days to Sales_Date.
from pyspark.sql.functions import date_sub 
df.select("Sales_Date",date_sub("Sales_date",-20).alias("subtract 20 days")).show()
```

```python
# Q13. Calculate the number of days between current_date() and Sales_Date.
from pyspark.sql.functions import datediff 
from pyspark.sql.functions import current_date
df.select("sales_date",datediff("Sales_Date",current_date()).alias("no.of days b/w current date and sales date")).show()
```

```python
# Q14. Calculate the number of months between current_date() and Sales_Date.
from pyspark.sql.functions import months_between 
df.select("Sales_Date",months_between("Sales_Date",current_date()).alias(" between months")).show()
```

```python
# Q15. Find the next Monday after Sales_Date.
from pyspark.sql.functions import next_day 
df.select("Sales_Date",next_day("Sales_Date","monday").alias("next monday")).show()
```

```python
# Q16. Convert Sales_Date into format dd-MM-yyyy.
from pyspark.sql.functions import date_format 
df.select('Sales_Date',date_format('Sales_Date','dd-mm-yyyy').alias('date format')).show()

df.select("Sales_Date",date_format("Sales_Date","mm-dd-yyyy").alias(" date format")).show()
```
