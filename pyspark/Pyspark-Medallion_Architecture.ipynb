```python
## 1 Load all SalesLT tables from Azure SQL into Bronze Delta tables
tables = {
    "address": df_address,
    "customer": df_customer,
    "customer_address": df_customer_address,
    "product": df_product,
    "product_category": df_product_category,
    "product_description": df_product_description,
    "product_model": df_product_model,
    "product_model_description": df_product_model_description,
    "sales_order_header": df_sales_order_header,
    "sales_order_detail": df_sales_order_detail
}

for table_name, df in tables.items():
    df.write.format("delta") \
      .mode("overwrite") \
      .saveAsTable(f"bronze.{table_name}")
```

```python
## 3
from pyspark.sql.functions import to_date
df_sales = spark.table("bronze.sales_order_header")
silver_sales = df_sales.withColumn(
    "order_date",
    to_date("OrderDate")
)
silver_sales.show()
```

```python
## 5
from pyspark.sql.functions import sum
df_sales_order_header = spark.table("bronze.sales_order_header")
df_customer = spark.table("bronze.customer")
df_customer_address = spark.table("bronze.customer_address")
df_address = spark.table("bronze.address")
sales_city = df_sales_order_header \
    .join(df_customer, "CustomerID") \
    .join(df_customer_address, "CustomerID") \
    .join(df_address, "AddressID") \
    .groupBy("City", "StateProvince") \
    .agg(
        sum("TotalDue").alias("sales")
    )
sales_city.show()
```

```python
## 18
from pyspark.sql.window import Window
from pyspark.sql.functions import rank
monthly = df_sales_header \
.groupBy(
    year("OrderDate").alias("year"),
    month("OrderDate").alias("month")
).agg(
    sum("TotalDue").alias("sales")
)
window_spec = Window.partitionBy(
    "year"
).orderBy(
    monthly["sales"].desc()
)

result = monthly.withColumn(
    "rank",
    rank().over(window_spec)
).filter("rank = 1")
result.show()
```
