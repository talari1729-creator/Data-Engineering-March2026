```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *

spark=SparkSession.builder.appName("data functions demo").getOrCreate()
# sample data
date_function=[
    (1, "2026-06-08", "2026-06-08 10:15:30"),
    (2, "2025-12-25", "2025-12-25 18:45:20"),
    (3, "2024-01-01", "2024-01-01 08:20:10")
]
df=spark.createDataFrame(date_function,["id","date_str","timestamp"])

df=df.withColumn("date_col",to_date("date_str")) \
        .withColumn("timestamp_col",to_timestamp("timestamp"))
```

```python
df.show()
```

```python
# 1. current_date()
df.select(current_date().alias("current date")).show()
```

```python
# 2. current_timestamp()
df.select(current_timestamp().alias("current timestamp")).show(truncate=False)
```

```python
# 3. year()
df.select("date_col",year("date_col").alias("year")).show()
```

```python
# 4. month()
df.select("date_col",month("date_col").alias("month")).show()
```

```python
# 5. dayofmonth()
df.select("date_col",dayofmonth("date_col").alias(" day of month")).show()
```

```python
# 6. dayofweek()
df.select("date_col",dayofweek("date_col").alias("day of week")).show()
```

```python
# 7. dayofyear()
df.select("date_col",dayofyear("date_col").alias("day of year")).show()
```

```python
# 8. weekofyear()
df.select("date_col",weekofyear("date_col").alias("week of year")).show()
```

```python
# 9. quarter()
df.select("date_col",quarter("date_col").alias("quarter")).show()
```

```python
# 10. last_day()
df.select("date_col",last_day("date_col").alias("last day")).show()
```

```python
# 11. add_months()
df.select("date_col",add_months("date_col",3).alias("addition of months")).show()
```

```python

# 12. date_add()
df.select("date_col",date_add("date_col",3).alias("date added")).show()
```

```python
# 13. date_sub()
df.select("date_col",date_sub("date_col",4).alias("date subtracted")).show()
```

```python
# 14. datediff()
df.select("date_col",datediff(current_date(),"date_col").alias(" days_iff")).show()
```

```python
# 15. months_between()
df.select("date_col",months_between(current_date(),"date_col").alias("days between")).show()
```

```python
# 16. next_day()
df.select("date_col",next_day("date_col","tuesday").alias("next tuesday date")).show()

```

```python
# 17. date_format()
df.select("date_col",date_format("date_col","dd-mm-yyyy").alias("date format")).show()
```

```python
# 18. hour()
df.select("timestamp_col",hour("timestamp_col").alias("hour")).show()
```

```python
# 19. minute()
df.select("timestamp_col",minute("timestamp_col").alias("minutes")).show()
```

```python
# 20. second()
df.select("timestamp_col",second("timestamp_col").alias("seconds")).show()
```
