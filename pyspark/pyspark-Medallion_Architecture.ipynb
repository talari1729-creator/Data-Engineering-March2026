```python
# 1.Create customer sales summary tables
from pyspark.sql.functions import sum
df_total_sales=df_sales_order_header.groupBy("customerID").agg(sum("totaldue").alias("total_sales")).show()
```

```python
# 2. create monthly sales 
df_monthly_sales=df_sales_order_header\
    .groupBy(
        year("orderdate").alias("year"),
        month("orderdate").alias("month")
    ).agg(sum("totaldue").alias("sales"))

df_monthly_sales.show()
```

```python
# 3.find top 10 customers by revenue 
from pyspark.sql.functions import desc
df_top_customers=df_sales_order_header\
    .groupBy("customerid")\
    .agg(sum("totaldue").alias("revenue"))\
    .orderBy(desc("revenue"))\
    .limit(10)

df_top_customers.show()
```

```python
# 4.find top 10 products by sales amount 
#df_sales_order_detail.select("linetotal").show()
#df_product.select("name").show()

df_top_products=df_product\
    .join(df_sales_order_detail,"productid")\
    .groupBy("name")\
    .agg(sum("linetotal").alias("sales"))\
    .orderBy(desc("sales"))\
    .limit(10)

df_top_products.show()


```

```python
# 5.find sales by city and state 
df_city_state=df_sales_order_header\
    .join(df_customer,"customerid")\
    .join(df_customer_address,"customerid")\
    .join(df_address,"addressid")\
    .groupBy("city","stateprovince")\
    .agg(sum("totaldue").alias("sales"))\
    .orderBy(desc("sales"))
df_city_state.show()
    
    
    
```

```python
# 6.category wise revenue report 
df_category_revenue=df_sales_order_detail\
    .join(df_product,"productid")\
    .join(df_product_category,"productcategoryid")\
    .groupBy(df_product_category["name"].alias("category_name"))\
    .agg(sum("linetotal").alias("revenue"))

df_category_revenue.show()
```

```python
# 7.total orders per customers 
from pyspark.sql.functions import countDistinct
df_orders_per_customer=df_sales_order_header\
    .join(df_customer,"customerid")\
    .groupBy("customerid")\
    .agg(countDistinct("salesorderid").alias("orders"))

df_orders_per_customer.show()
    
```
