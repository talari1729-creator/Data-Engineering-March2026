```python
## SET Types----mutable...using symbol ( {} )
## set types represents unordered collection of unique items/hashable items.
## They are useful for membership testing,removing duplicates and performing mathematical set of oerations(like union,intersection,difference).
## hashable means items must be immutable(like numbers,strings,tuple,but not list or other sets.)

```

```python
## creating a set and add items(ignores duplicates)
my_info={"name","age","place","city","name","age"}
print(my_info)
```

```python
## create a set using() from  a list (removes duplicates)
my_family=["amma","nanna","nanna","chinnodu","peddodu","peddodu"]
set_type=set(my_family)
print(set_type)


my_set=set()
print(my_set)  ## creating a empty set
```

```python
## adding items to the empty list
my_set
my_set=("name")
print(my_set)
my_set=("nanna")
print(my_set)
```
