```python
## class method
##  method work with the class level data (class attributes) not instance attributes
## it uses the decorator (@classmethod) and receives the cls as 1st parameter
## it does not depend on object

class cart:
    flat_discount=0
    @classmethod
    def update_discount(cls,new_discount):
        cls.flat_discount=new_discount
```

```python
cart.update_discount(222)
```

```python
cart.flat_discount
```

```python
## multiple class methods

class cart:
    flat_discount=0
    @classmethod
    def update_discount(cls,new_discount):
        cls.flat_discount=new_discount
        ## Another class method for increment the discount 
    @classmethod
    def increment_discount(cls,amount):
        new_discount=cls.flat_discount+amount
        cls.update_discount(new_discount)
```

```python
## without object and instances
cart.update_discount(250)
```

```python
cart.flat_discount
```

```python
cart.increment_discount(250)
```

```python
cart.flat_discount
```

```python
## static method 
## this method not access the both instance and class data
## it behave like a normal function but placed inside a class for logocal grouping
# this method uses the @staticmethod decorator

class cart:
    @staticmethod
    def greet():
        print(f"hi buchi how are you")
        
```

```python
## instance VS class VS static
class employee:
    company="google"

    def __init__(self,name):
        self.name=name

    def show_name(self):
        print(f"name of the employee:",self.name)
    @classmethod
    def show_company(cls):
        print(f"company name:",cls.company)

    @staticmethod
    def greet(employee):
        print(f"welcome: Mr.",employee.name)
```

```python
emp=employee("raajan")
emp.show_name()
emp.show_company()
employee.greet(emp)
```

```python
class library:
    def __init__(self):
        self.books={}
    def add_books(self,book_name,author,publisher,pen_name,published_year):
        self.books[book_name]={
            "author":author,
            "publisher":publisher,
            "pen_name":pen_name,
            "published_year":published_year
            }
    def display_books(self):
        for book_name,details in self.books.items():
            print(f"the book name is:{book_name}")
            print(f"the author name is:{details['author']}")
            print(f"the publisher is:{details['publisher']}")
            print(f"the pen name is:{details['pen_name']}")
            print(f"published year is:{details['published_year']}")
            print()

    def remove_books(self,book_name):
        if book_name in self.books:
            del self.books[book_name]
            print(f"{book_name} removed successfully")
        else:
            print(f"book not found")

    def update_books(self,book_name,author,publisher,pen_name,published_year):
        if book_name in self.books:
            self.books[book_name]["author"]=author
            self.books[book_name]["publisher"]=publisher
            self.books[book_name]["pen_name"]=pen_name
            self.books[book_name]["published_year"]=published_year
            print(f"successfully updated")
        else:
            print(f"book not found")

    def find_books(self,book_name):
        if book_name in self.books:
            self.books[book_name]
            print(f"{book_name} found successfully")
        else:
            print(f"book not found")

    def replace_books(self,old_book,new_book):
        if old_book in self.books:
            self.books[new_book]=self.books.pop(old_book)
            print(f"{old_book} is successfully replaced with {new_book}")
        else:
            print(f"book not found")
       
```

```python
class cart:
    def __init__(self):
        self.items={}

    def add_items(self,item_name,brand,quantity,price):
        self.items[item_name]={
            "brand":brand,
            "quantity":quantity,
            "price":price
        }
    def display_items(self):
        for item_name,details in self.items.items():
            print(f"item name:{item_name}")
            print(f"brand name :{details['brand']}")
            print(f"quantity:{details['quantity']}")
            print(f"price of item:{details['price']}")
            print()
        
    def update_items(self,item_name,quantity,brand,price):
       if item_name in self.items:
           self.items[item_name]["brand"]=brand
           self.items[item_name]["quantity"]=quantity
           self.items[item_name]["price"]=price
           print(f"{item_name} updated successfully")
       else:
           print("item not found")
      
    def remove_items(self,item_name):
        if item_name in self.items:
            del self.items[item_name]
            print(f"{item_name} removed successfully")
        else:
            print("item not found")
        
    def search_items(self,item_name):
        if item_name in self.items:
            self.items[item_name]
            print(f"{item_name} search successfully")
        else:
            print("item not found")
            
    def replace_items(self, old_name, new_name):
        if old_name in self.items:
            self.items[new_name] = self.items.pop(old_name)
            print(f"{old_name} replaced with {new_name} successfully")
        else:
            print("Item not found")
```

```python
class bankbalance:
   

    def __init__(self,ac_holder,ac_number,branch,balance):
        self.ac_holder=ac_holder
        self.ac_number=ac_number
        self.branch=branch
        self.balance=balance

    def display(self):
        print(f"account holder name is :",{self.ac_holder})
        print(f"account number of customer:",{self.ac_number})
        print(f"branch of the bank :",{self.branch})
        print(f"bank balance is:",{self.balance})



    def deposite(self,amount):
       
        self.balance+=amount
        print(f"your current balance is :",{self.balance})

    def withdraw(self,amount):
      
        if amount > self.balance:
            print(f"insufficient balance")
        else:
           self.balance -= amount
           print(f"withdraw success")

    def check_balance(self):
        print(f"present balance:",self.balance)
```
