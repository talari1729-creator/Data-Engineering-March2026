## floor division
## dividies the first number by the second number and result should be round to the down nearest whole number

## modulo % 
## it returns the remainder values while the first number divide by the second number

```python
## finding remain_seats
total_customers=199
total_marts=17
remain_customers=total_customers % total_marts
print(remain_customers)
```

```python
## finding left over assignments
assignments=123
instructors= 14
remain_assignments=assignments % instructors
print(remain_assignments)
```

```python
## checking for even or odd
student_id=1729
is_odd=student_id % 2
print(is_odd)
print(f"is student id {student_id} odd (1 means yes)? {is_odd}")
```
