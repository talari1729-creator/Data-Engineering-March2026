```python
# Question 2:find the real part of the complex 7-2j
growth_projection=7-2j
real_part=growth_projection.real
print(real_part)
```

```python
# Question 3:find the imaginary part the complex
growth_projection=4+6j
imaginary_part=growth_projection.imag
print(imaginary_part)
```

```python
# Question 4:add two complex numbers (3+2j) and (1+4j)
c1=3+2j
c2=1+4j
add_complex=c1+c2
print(add_complex)
```

```python
# Question 5:Multiply two complex numbers
mc1=2+3j
mc2=1+2j                   ## (a+bj)(c+dj) = (ac-bd) + (ad+bc)j
complex_multi=mc1*mc2
print(complex_multi)
```

```python
# Question 6:find the magnitude of the complex
complex1=6+8j            # ex: sqrt(10^2 + 3^2) = sqrt(100 + 9) = sqrt(109) approx 10.44
mangni_complex=abs(complex1)
print(mangni_complex)
```

```python
# Question 7: find the conjugate of complex number
complex_conju=5_7j
conjugate_=complex_conju.conjugate()
print(conjugate_)
```

```python
# Question 8:subtract complex numbers
comp_1=10+5j
comp_2=3+2j
sub_comp=comp_1-comp_2
print(sub_comp)
```

```python
# Question 9:divide the complex numbers
com_1=8+6j
com_2=2+1j
div_com=com_1/com_2
print(div_com)
```

```python
# Question 10:find the phase angle of complex numbers
import cmath

z = 1 + 1j
theta = cmath.phase(z)   # radians

print("Phase angle (radians):", theta)
print("Phase angle (degrees):", theta * 180 / cmath.pi)
```
