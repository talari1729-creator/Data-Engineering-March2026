```python
## 19. Merge Dictionaries with Cumulative Value      ## output {'a': 10, 'b': 50, 'c': 40}
dict1 = {'a': 10, 'b': 20}
dict2 = {'b': 30, 'c': 40}

def merge_dicts(dict1, dict2):
    result = dict1.copy()

    for key, value in dict2.items():
        if key in result:
            result[key] += value
        else:
            result[key] = value

    return result


dict1 = {'a': 10, 'b': 20}
dict2 = {'b': 30, 'c': 40}

print(merge_dicts(dict1, dict2))
```

```python
## 25. Factorial Calculation                           ## factorial of 5 using recursive function

def facto(n:int):
    if n!=1:
        return n * facto(n-1)
    else:
        return 1
print(facto(5))
```

```python
n=5
facto=1
for i in range(1,n+1):
    facto*=i
print(facto)
```

```python
def fibonacci(n):
    if n>=1:
        return fibonacci(n-1)+fibonacci(n-2)
    else:
        return n

n=6

print(fibonacci(n))
```

```python
## 28. Reverse a String Recursively
str="python"
def reverse(string):
    if len(string)==0:
        return string
    return reverse(string[1:])+string[0]
text="python"
print(reverse(text))
```
