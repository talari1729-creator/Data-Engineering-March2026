```python
set_a={1,2,3,4}
set_b={3,4,5,6}
result=set_a.union(set_b)
print(result)
set_c={7,8,9}
set_c

## multiple of sets
results=set_a|set_b|set_c
print(results)
```

```python
## interseection ---
## return common item only
## return the item that exists in both the sets

set_A={1,2,3,4}
set_B={3,4,5,6}
result=set_A.intersection(set_B)
print(result)

## intersection also read the symbol of (&)
result_1=set_A & set_B
print(result_1)
```

```python
## difference - item in the first set but not in the second set
## vice versa
set_aa={1,2,3,4}
set_bb={3,4,5,6}

result_1=set_aa.difference(set_bb)
print(result_1)

result_2=set_aa - set_bb  ## for diff use (-)
print(result_2)

result_3=set_bb - set_aa
print(result_3)
```

```python
## symmetric difference - item in either set but not both
## it avoids the common elements
## using the symbol (^) and .symmetric.difference()


set_AA={1,2,3,4}
set_BB={3,4,5,6}
result= set_AA ^ set_BB
print(result)
```

```python
## set comparation operations

## subset - is every item of set a in set b

## that means set a is subset of set b--viceversa
set_AA={4,5,6,7}
set_BB={4,5,6,7,8}
```

```python
## check if set_AA IS subset of set_BB
print(set_AA.issubset(set_BB))

print(set_BB.issubset(set_AA))
```

```python
## disjoint - sets does not have common items
set_11={1,2,3}
set_22={4,5,6}
set_33={3,6,7}
print(set_11.isdisjoint(set_22))
print(set_22.isdisjoint(set_33))
```

```python
def calculate_average(scores):
    """this function takes list of numbers and returns the average of numbers"""
    total=sum(scores)
    avg=total/len(scores)
    return avg

print(calculate_average([12,23,34]))
```
