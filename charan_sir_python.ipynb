```python
str1="virat kohli jersy no 20"

digit=0
alpha=0
special=0

for ch in str1:
    if ch >'0' and ch<'9':
        digit+=1
    elif (ch>'a' and ch<'z')or(ch>'A' and ch<'Z'):
        alpha+=1
    else:
        special+=1

print(digit,alpha,special)
```
