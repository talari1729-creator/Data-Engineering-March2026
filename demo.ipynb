```python
class cart:
    flat_discount=0

    @classmethod
    def update_flat_discount(cls,new_flat_discount):
        cls.flat_discount=new_flat_discount

    @classmethod
    def increase_flat_discount(cls,amount):
        new_flat_discount=cls.flat_discount+amount
        cls.update_flat_discount(new_flat_discount)

cart.update_flat_discount(10000)

cart.increase_flat_discount(555)

print(cart.flat_discount)


```

```python
class cart:
    def __init__(self):
        self.items={}

    def add_items(self,item_name,brand,quantity,price):
        self.items[item_name]={
            "brand":brand,
            "quantity":quantity,
            "price":price
        }
    def display_items(self):
        for item_name,details in self.items.items():
            print(f"item name:{item_name}")
            print(f"brand name :{details['brand']}")
            print(f"quantity:{details['quantity']}")
            print(f"price of item:{details['price']}")
            print()
        
    def update_items(self,item_name,quantity,brand,price):
       if item_name in self.items:
           self.items[item_name]["brand"]=brand
           self.items[item_name]["quantity"]=quantity
           self.items[item_name]["price"]=price
           print(f"{item_name} updated successfully")
       else:
           print("item not found")
      
    def remove_items(self,item_name):
        if item_name in self.items:
            del self.items[item_name]
            print(f"{item_name} removed successfully")
        else:
            print("item not found")
        
    def search_items(self,item_name):
        if item_name in self.items:
            self.items[item_name]
            print(f"{item_name} search successfully")
        else:
            print("item not found")
            
    def replace_items(self, old_name, new_name):
        if old_name in self.items:
            self.items[new_name] = self.items.pop(old_name)
            print(f"{old_name} replaced with {new_name} successfully")
        else:
            print("Item not found")
            

```

```python
class bankbalance:
   

    def __init__(self,ac_holder,ac_number,branch,balance):
        self.ac_holder=ac_holder
        self.ac_number=ac_number
        self.branch=branch
        self.balance=balance

    def display(self):
        print(f"account holder name is :",{self.ac_holder})
        print(f"account number of customer:",{self.ac_number})
        print(f"branch of the bank :",{self.branch})
        print(f"bank balance is:",{self.balance})



    def deposite(self,amount):
       
        self.balance+=amount
        print(f"your current balance is :",{self.balance})

    def withdraw(self,amount):
      
        if amount > self.balance:
            print(f"insufficient balance")
        else:
           self.balance -= amount
           

    def check_balance(self):
        print(f"present balance:",self.balance)

   

```
